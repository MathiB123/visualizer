## Section 2.2 - Exemple de série de Fourier avec la fonction dents de scie

```python
import numpy as np
import matplotlib.pyplot as plt
```

La fonction dents de scie

```python
A = 1
L = 2

n_points = 100   #Le nombre de points
x_s = np.linspace(-L/2, L/2, n_points)
y_s = A * x_s

plt.plot(x_s, y_s, color="blue", label="Fonction dents de scie")
plt.plot(x_s + L, y_s, color="blue")
plt.plot(x_s - L, y_s, color="blue")
plt.xlabel("x")
plt.ylabel(f"f(x) = Ax sur {-L/2} à {L/2}, période de {L}")
plt.grid()
plt.legend()
plt.show()
```

Série de Fourier pour la fonction dents de scie

```python
N = 5  #Le nombre de termes dans la somme

#Testez avec N=5, N=50 et N=75!

y_fourier = np.zeros(x_s.shape)

for i in range(N):
    y_fourier += (A*L/np.pi) * (-1)**(i+2) * 1/(i+1) * np.sin(2*np.pi*(i+1)*x_s*(1/L))

plt.plot(x_s, y_s, color="blue", label="Fonction dents de scie")
plt.plot(x_s + L, y_s, color="blue")
plt.plot(x_s - L, y_s, color="blue")
plt.plot(x_s, y_fourier, color="green")
plt.plot(x_s + L, y_fourier, color="green")
plt.plot(x_s - L, y_fourier, color="green")
plt.xlabel("x")
plt.ylabel(f"f(x) = Ax sur {-L/2} à {L/2}, période de {L}")
plt.grid()
plt.legend()
plt.show()
```

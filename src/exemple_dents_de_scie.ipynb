### Section 2.2 - Exemple de série de Fourier avec la fonction dents de scie

```python
import numpy as np
import matplotlib.pyplot as plt
```

La fonction dents de scie

```python
A = 2
T = 2

n_points = 1000   #Le nombre de points
x_s = np.linspace(-T/2, T/2, n_points)
y_s = A * x_s

plt.plot(x_s, y_s, color="blue", label="Fonction dents de scie")
plt.plot(x_s + T, y_s, color="blue")
plt.plot(x_s - T, y_s, color="blue")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.legend()
plt.show()
```

Série de Fourier pour la fonction dents de scie

```python
N = 5 #Le nombre de termes dans la somme

#Testez avec N=5, N=25 et N=50!

x_fourier = np.linspace(-3*T/2, 3*T/2, n_points)
y_fourier = np.zeros(x_s.shape)

for i in range(N):
    y_fourier += (A*T/np.pi) * (-1)**(i+2) * 1/(i+1) * np.sin(2*np.pi*(i+1)*x_fourier*(1/T))

plt.plot(x_s, y_s, color="blue", label="Fonction dents de scie")
plt.plot(x_s + T, y_s, color="blue")
plt.plot(x_s - T, y_s, color="blue")
plt.plot(x_fourier, y_fourier, color="green", label="Série de Fourier")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.legend()
plt.show()
```

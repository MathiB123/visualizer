```python
from scipy.integrate import quad
import numpy as np
import matplotlib.pyplot as plt
```

```python
tau = 1
num_points = 1000
x_s = np.linspace(-tau, tau, num_points)
rect = np.zeros_like(x_s)

#Bornes d'intégration
a = -500
b = -a

def Re(tau, omega, x):
    return np.sin(tau*omega/2)*np.cos(omega*x)/omega

for i,x in enumerate(x_s):
    rect[i] = quad(lambda omega: Re(tau, omega, x), a, b, points=[0])[0]/np.pi
```

```python
plt.plot(x_s, rect)
plt.xlabel("x")
plt.ylabel("Rect(x)")
plt.grid()
plt.show()
```
